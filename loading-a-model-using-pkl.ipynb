```python
import pickle
import sklearn
import pandas as pd
```

```python
fp = open("./models/dtc.pkl", "rb")
model = pickle.load(fp)
fp.close()
```

```python
df = pd.read_csv("./models/dtc-testing-samples.csv")

X = df.drop(columns=["diseases"])
y = df["diseases"]
```

```python
predictions = model.predict(X)
predictions
```

```python
accuracy = sklearn.metrics.accuracy_score(y.values, predictions)
accuracy
```
