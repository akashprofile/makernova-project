```python
import pandas as pd
```

```python
original = pd.read_csv("./FilteredDataset.csv")
original.shape
```

```python
partition1 = original.iloc[0:60000, ::]
partition1.shape
```

```python
partition2 = original.iloc[60000:120000:, ::]
partition2.shape
```

```python
partition3 = original.iloc[120000:180000, ::]
partition3.shape
```

```python
partition4 = original.iloc[180000::, ::]
partition4.shape
```

```python
# check
partition1.shape[0] + partition2.shape[0] + partition3.shape[0] + partition4.shape[0]
```

```python
partition1.to_csv("./dataset-partition-1.csv", index=False)
partition2.to_csv("./dataset-partition-2.csv", index=False)
partition3.to_csv("./dataset-partition-3.csv", index=False)
partition4.to_csv("./dataset-partition-4.csv", index=False)
```
