```python
import pandas as pd

# NOTE: Original dataframe had 246945 entries

partition1 = pd.read_csv("./dataset-partition-1.csv")
partition2 = pd.read_csv("./dataset-partition-2.csv")
partition3 = pd.read_csv("./dataset-partition-3.csv")
partition4 = pd.read_csv("./dataset-partition-4.csv")
```

```python
# combining the partitions to get original one.
df = pd.concat([partition1, partition2, partition3, partition4])
df.info(verbose=True, show_counts=True)
```

```python
# retaining features matrix only
df.drop(columns=["Unnamed: 0", "diseases"], inplace=True)
```

```python
# getting the symptoms as list
symptoms = df.columns.to_list()

# creating a dictionary whose keys represent columns ans value is an empty array, those will serve as placeholders for alternative text
symptoms = { symptom: [] for symptom in symptoms }

# exporting this dictionary as JSON file
import json 
with open("./alternative-texts-duplicate.json", "w") as json_file:
  json.dump(symptoms, json_file, indent=4)

# NOTE: chnaged output file name to prevent changing the original file by mistake.
```
